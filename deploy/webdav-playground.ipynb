```python
import os

import requests

from webdav3.client import Client, RemoteResourceNotFound
```

```python
options = {
    "webdav_hostname": "http://localhost:8081",
    "webdav_login": "weber",
    "webdav_password": "1234",
    "webdav_timeout": 30,
}

client = Client(options)
```

```python
PROPFIND_REQUEST = """<?xml version="1.0" encoding="utf-8" ?>
<d:propfind xmlns:d="DAV:">
<d:prop xmlns:oc="http://owncloud.org/ns">
  <d:getlastmodified/>
  <oc:size/> 
</d:prop>
</d:propfind>"""
```

```python
s = requests.Session()
s.auth = ("weber", "1234")
r = s.request(
    method="PROPFIND",
    url=f"{options['webdav_hostname']}/adhoc/testtest",
    data=PROPFIND_REQUEST,
    headers={"Depth": "0"},
)
```

```python
head = "adhoc/testtest/2023-09-22-test&break&me/"
path_pyramid = []

while True:
    head, tail = os.path.split(head)
    # stripped trailing slash
    if not tail and head != "/":
        continue
    # no leading slash and no separator
    if tail:
        path_pyramid.insert(0, (head, tail))
    else:
        break
    if head == "":
        break
```
