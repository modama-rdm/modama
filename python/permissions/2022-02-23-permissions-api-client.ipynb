```python
import requests
```

```python
base_url = "http://localhost:5050"
```

# Authenticate

```python
r = requests.post(
    f"{base_url}/users/auth/", json={
        "username": "testdieter",
        "password": "1234",
    }
)
print(r)
r.json()
```

# List projects

```python
requests.get(f"{base_url}/project/").json()
```

# Create new adhoc project

```python
r = requests.post(
    f"{base_url}/project/", json={
        "name": "testproject",
        "owner": "testdieter",
        "members": ["testdieter", "testalex"]
    }
)
print(r)
r.json()
```

# Get project details

```python
requests.get(f"{base_url}/project/testproject/").json()
```

# Create dataset

```python
requests.post(
    f"{base_url}/project/testproject/dataset/", json={
        "name": "testds3",
        "owner": "testdieter",
    }
).json()
```

# List datasets

```python
requests.get(f"{base_url}/project/testproject/dataset/").json()
```

# Seal dataset

```python
requests.post(
    f"{base_url}/project/testproject/dataset/seal/", json={
        "name": "testds2",
        "user": "testdieter",
    }
).json()
```

```python
requests.get(f"{base_url}/project/testproject/dataset/").json()
```

# Change project membership

```python
requests.patch(
    f"{base_url}/project/testproject/members/",
    json={
        "op": "add",
        "members": ["testpenghan"],
        "user": "testdieter"
    },
).json()
```

```python
requests.patch(
    f"{base_url}/project/testproject/members/",
    json={
        "op": "remove",
        "members": ["testpenghan"],
        "user": "testpenghan"
    },
).json()
```
