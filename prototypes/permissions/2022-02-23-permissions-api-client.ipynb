```python
import requests
```

```python
base_url = "http://172.28.0.2:5050"
```

# List projects

```python
requests.get(f"{base_url}/project/").json()
```

# Create new adhoc project

```python
requests.post(
    f"{base_url}/project/", json={
        "name": "testproject",
        "owner": "testdieter",
        "members": ["testdieter", "testalex"]
    }
)
```

# Get project details

```python
requests.get(f"{base_url}/project/testproject/").json()
```

# Create dataset

```python
requests.post(
    f"{base_url}/project/testproject/datasets/", json={
        "name": "testds",
        "owner": "testdieter",
    }
).json()
```

# List datasets

```python
requests.get(f"{base_url}/project/testproject/datasets/").json()
```

# Seal dataset

```python
requests.post(
    f"{base_url}/project/testproject/datasets/seal/", json={
        "name": "testds",
    }
).json()
```
