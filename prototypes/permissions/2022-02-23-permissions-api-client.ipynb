```python
import requests
```

```python
base_url = "http://172.28.0.2:5050"
```

# List projects

```python
requests.get(f"{base_url}/project/").json()
```

# Create new adhoc project

```python
r = requests.post(
    f"{base_url}/project/", json={
        "name": "testproject",
        "owner": "testdieter",
        "members": ["testdieter", "testalex"]
    }
)
print(r)
r.json()
```

# Get project details

```python
requests.get(f"{base_url}/project/testproject/").json()
```

# Create dataset

```python
requests.post(
    f"{base_url}/project/testproject/dataset/", json={
        "name": "testds2",
        "owner": "testdieter",
    }
).json()
```

# List datasets

```python
requests.get(f"{base_url}/project/testproject/dataset/").json()
```

# Seal dataset

```python
requests.post(
    f"{base_url}/project/testproject/dataset/seal/", json={
        "name": "testds",
    }
).json()
```

```python
requests.get(f"{base_url}/project/testproject/dataset/").json()
```

# Change project membership

```python
requests.patch(
    f"{base_url}/project/testproject/members/",
    json={
        "op": "add",
        "members": ["testpenghan"],
    },
).json()
```

```python
requests.patch(
    f"{base_url}/project/testproject/members/",
    json={
        "op": "remove",
        "members": ["testpenghan"],
    },
).json()
```
