```python
import win32net
```

```python
def get_drives():
    result = []
    resume = 0
    while 1:
        (drives, _total, resume) = win32net.NetUseEnum(None, 0, resume)
        for drive in drives:
            if drive["local"]:
                _, _, server, share = drive["remote"].split("\\")
                result.append((drive["local"], server, share))
        if not resume:
            break
    return result
```

```python
get_drives()
```
