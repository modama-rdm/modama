```python
import os

from webdav3.client import Client
```

```python
options = {
    'webdav_hostname': "http://localhost:8081",
    'webdav_login':    "weber",
    'webdav_password': "1234",
    'webdav_timeout': 30
}

client = Client(options)
    
```

```python
client.list('adhoc/testtest/2023-09-22-test&break&me/', get_info=True)
```

```python
client.check('adhoc/testtest/2023-09-22-test&break&me/')
```

```python
head = 'adhoc/testtest/2023-09-22-test&break&me/'
path_pyramid = []

while True:
    head, tail = os.path.split(head)
    # stripped trailing slash
    if not tail and head != '/':
        continue
    # no leading slash and no separator
    if tail:
        path_pyramid.insert(0, (head, tail))
    else:
        break
    if head == '':
        break
```
